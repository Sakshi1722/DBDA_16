```
import re

def preprocess_caption(i):
    caption = i.lower()  # Convert to lowercase
    caption = re.sub(r'[^\w\s]', '', caption)  # Remove punctuation
    caption = re.sub(r'\d+', '', caption)  # Remove numbers
    words = caption.split()  # Tokenize (split into words)
    words = ['<start>'] + words + ['<end>']  # Add special tokens
    return words
# Apply preprocessing to all captions
preprocessed_captions = {}
for img, captions in captions_dict.items():
    preprocessed_captions[img] = [preprocess_caption(caption) for caption in captions]

```

```
for k,v in preprocessed_captions.items():
  print(k,v)
```

```
# Build vocabulary from tokenized captions
all_words = set(word for captions in preprocessed_captions.values() for caption in captions for word in caption)

# Map words to indices
word_to_index = {word: idx for idx, word in enumerate(all_words, start=1)}
word_to_index['<PAD>'] = 0  # Include a special <PAD> token
```

```
# Convert tokenized words to numbers
numeric_captions = {}
for img, captions in preprocessed_captions.items():
    numeric_captions[img] = [[word_to_index[word] for word in caption] for caption in captions]
```

```
# Example: Save numeric captions
numeric_captions = {img: [[word_to_index[word] for word in caption] for caption in captions]
                    for img, captions in preprocessed_captions.items()}
for k,v in numeric_captions.items():
  print(k,v)
print(type(numeric_captions))
```

```
import numpy as np
from tensorflow.keras.preprocessing.sequence import pad_sequences

# Create training data **without padding first**
X_train = []
y_train = []

for captions in numeric_captions.values():
    for caption in captions:
        for i in range(1, len(caption)):  # Generate input-output pairs
            X_train.append(caption[:i])  # Input sequence (partial caption)
            y_train.append(caption[i])   # Target (next word)

# Determine max length
max_len = max(len(seq) for seq in X_train)

# **Pad only X_train before feeding into the model**
X_train = pad_sequences(X_train, maxlen=max_len, padding='pre')
y_train = np.array(y_train)

print("X_train shape:", X_train.shape)
print("y_train shape:", y_train.shape)
```

```
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense

all_numeric_captions = list(numeric_captions.values())
all_numeric_captions = [seq for sublist in all_numeric_captions for seq in sublist]
vocab_size = max([max(seq) for seq in all_numeric_captions]) + 1

#vocab_size = len(numeric_captions)  # Vocabulary size from numeric conversion

model = Sequential([
    Embedding(input_dim=vocab_size, output_dim=256, mask_zero=True),  # Word embeddings
    LSTM(256, return_sequences=False),  # LSTM layer
    Dense(vocab_size, activation='softmax')  # Output layer (predict next word)
])

model.compile(loss='sparse_categorical_crossentropy', optimizer='adam', metrics=['accuracy'])


# Train the model
model.fit(X_train, y_train, epochs=10, batch_size=64)
```
